```python
from langchain.messages import HumanMessage, AIMessage

scifi_agent = create_agent(
    model=get_model(),
    system_prompt="You are a science fiction writer. Reply ONLY with an invented capital city name.",
)

response = scifi_agent.invoke({"messages": [
    HumanMessage("What is the capital of Mars?"),   AIMessage("Marsialis"),
    HumanMessage("What is the capital of Venus?"),  AIMessage("Venusovia"),
    question,
]})
print(response["messages"][-1].content)
```

## Structured output

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(
    model=get_model(),
    system_prompt="You are a science fiction writer, create a capital city at the users request.",
    response_format=CapitalInfo
)

question = HumanMessage(content="What is the capital of The Moon?")

response = agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
```

```python
### este modelo no soporta llamado de tools nno usa tools 
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(
    model=get_model("gemma3:4b"), #ResponseError: registry.ollama.ai/library/gemma3:4b does not support tools (status code: 400) During task with name 'model' and id '9e84bc5b-3d63-8700-dbf7-b139bc3c0944'
    system_prompt="You are a science fiction writer, create a capital city at the users request.",
    response_format=CapitalInfo
)

question = HumanMessage(content="What is the capital of The Moon?")

response = agent.invoke(
    {"messages": [question]}
)

response["structured_response"]
```
